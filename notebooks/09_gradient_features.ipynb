```python
import lasio
import numpy as np
import pandas as pd

RAW_DIR = PROJECT_ROOT / "data" / "raw" / "FORCE_2020"

MODEL_CURVES = ["GR", "RDEP", "RMED", "DTC", "RHOB"]
TARGET_COL = "FORCE_2020_LITHOFACIES_LITHOLOGY"

depth_parts = []
count_check = []

for i, las_path in enumerate(sorted(RAW_DIR.glob("*.las")), start=1):

    las = lasio.read(las_path)

    # ---------------------------------------------------------
    # Read raw LAS data
    # ---------------------------------------------------------
    raw = las.df().reset_index()

    # Normalize column names
    raw.columns = [str(c).upper() for c in raw.columns]

    if "DEPT" not in raw.columns:
        raise ValueError(f"DEPT not found in {las_path.name}")

    # ---------------------------------------------------------
    # Handle missing RMED exactly as preprocessing did
    # ---------------------------------------------------------
    if "RMED" not in raw.columns:
        raw["RMED"] = np.nan

    # Check remaining required columns
    required = [
        "DEPT",
        "GR",
        "RDEP",
        "RMED",
        "DTC",
        "RHOB",
        TARGET_COL,
    ]

    missing = [c for c in required if c not in raw.columns]

    if missing:
        raise ValueError(
            f"{las_path.name}: missing required columns: {missing}"
        )

    # ---------------------------------------------------------
    # Convert LAS NULL values to NaN
    # ---------------------------------------------------------
    raw = raw.replace([-999.25, -999.250000], np.nan)

    # ---------------------------------------------------------
    # SAME cleaning rules as established preprocessing
    # ---------------------------------------------------------

    # GR
    raw.loc[raw["GR"] < 0, "GR"] = np.nan

    # Resistivity
    raw.loc[raw["RDEP"] <= 0, "RDEP"] = np.nan
    raw.loc[raw["RMED"] <= 0, "RMED"] = np.nan

    # Sonic
    raw.loc[raw["DTC"] <= 0, "DTC"] = np.nan

    # Density
    raw.loc[raw["RHOB"] <= 0, "RHOB"] = np.nan

    # ---------------------------------------------------------
    # Same row-selection logic
    # ---------------------------------------------------------

    has_any_feature = raw[MODEL_CURVES].notna().any(axis=1)
    has_label = raw[TARGET_COL].notna()

    keep = has_any_feature & has_label

    selected = raw.loc[
        keep,
        ["DEPT"] + MODEL_CURVES + [TARGET_COL]
    ].copy()

    # ---------------------------------------------------------
    # Recover exact well name
    # ---------------------------------------------------------
    well_name = las_path.stem.replace("_", "/", 1)

    selected["WELL"] = well_name

    depth_parts.append(
        selected[["WELL", "DEPT"]]
    )

    count_check.append({
        "WELL": well_name,
        "raw_rows": len(raw),
        "selected_rows": len(selected),
        "RMED_available": "RMED" in las.df().columns,
    })

    if i % 20 == 0 or i == 118:
        print(f"Processed {i}/118 wells")


# -------------------------------------------------------------
# Combine all wells
# -------------------------------------------------------------

depth_recovered = pd.concat(
    depth_parts,
    ignore_index=True
)

count_check = pd.DataFrame(count_check)


# -------------------------------------------------------------
# Summary
# -------------------------------------------------------------

print("\n==============================================")
print("DEPTH RECONSTRUCTION SUMMARY")
print("==============================================")

print("Recovered rows :", len(depth_recovered))
print("Processed rows :", len(df))

print(
    "Row difference :",
    len(depth_recovered) - len(df)
)

print(
    "\nRecovered wells:",
    depth_recovered["WELL"].nunique()
)

print(
    "Processed wells:",
    df["WELL"].nunique()
)

print("\nFirst recovered rows:")
display(depth_recovered.head())

print("\nRMED availability:")
display(
    count_check["RMED_available"].value_counts()
)

print("\nPer-well count check:")
display(count_check.head)
```

```python
# ============================================================
# CELL 6 — ALIGN RECOVERED DEPTH WITH PROCESSED DATA
# ============================================================

# Our processed dataset uses the LAS filename convention:
# 15_9-13, 15_9-14, etc.
#
# The recovered dataframe currently contains:
# 15/9-13, 15/9-14, etc.
#
# Convert the recovered names back to the processed convention.

depth_recovered["WELL"] = (
    depth_recovered["WELL"]
    .astype(str)
    .str.replace("/", "_", regex=False)
)

# ------------------------------------------------------------
# Basic checks
# ------------------------------------------------------------

print("Processed wells :", df["WELL"].nunique())
print("Recovered wells :", depth_recovered["WELL"].nunique())

print("\nProcessed first wells:")
print(df["WELL"].drop_duplicates().head(10).tolist())

print("\nRecovered first wells:")
print(depth_recovered["WELL"].drop_duplicates().head(10).tolist())


# ------------------------------------------------------------
# Check whether the WELL sequence matches row-by-row
# ------------------------------------------------------------

processed_wells = df["WELL"].reset_index(drop=True)
recovered_wells = depth_recovered["WELL"].reset_index(drop=True)

well_alignment = processed_wells.equals(recovered_wells)

print("\n==============================================")
print("WELL ALIGNMENT CHECK")
print("==============================================")
print("Row count:", len(df))
print("WELL sequence matches:", well_alignment)


# ------------------------------------------------------------
# If it does not match, show the first mismatch
# ------------------------------------------------------------

if not well_alignment:

    mismatch = np.flatnonzero(
        processed_wells.to_numpy() != recovered_wells.to_numpy()
    )

    print("\nNumber of mismatched rows:", len(mismatch))

    if len(mismatch) > 0:
        idx = mismatch[0]

        print("\nFirst mismatch:")
        print("Row index:", idx)
        print("Processed WELL:", processed_wells.iloc[idx])
        print("Recovered WELL:", recovered_wells.iloc[idx])

        print("\nProcessed context:")
        display(
            df.iloc[max(0, idx - 3):idx + 4][["WELL"]]
        )

        print("\nRecovered context:")
        display(
            depth_recovered.iloc[max(0, idx - 3):idx + 4][["WELL", "DEPT"]]
        )

        raise ValueError(
            "WELL sequence still does not align. "
            "Do not calculate gradients yet."
        )

print("\n✓ WELL sequence alignment passed.")
```

```python
# ============================================================
# CELL 7 — ATTACH VERIFIED DEPTH TO ENGINEERED FEATURES
# ============================================================

# Make a copy so the original engineered dataset remains untouched.
df_v2 = df.copy()

# Add the recovered depth.
df_v2["DEPT"] = depth_recovered["DEPT"].to_numpy()

# ------------------------------------------------------------
# Check the resulting dataset
# ------------------------------------------------------------

print("Original engineered shape :", df.shape)
print("New feature-set shape    :", df_v2.shape)

print("\nDEPT missing values:")
print(df_v2["DEPT"].isna().sum())

print("\nDEPT data type:")
print(df_v2["DEPT"].dtype)

# ------------------------------------------------------------
# Check depth range for the whole dataset
# ------------------------------------------------------------

print("\nOverall depth range:")
print("Minimum:", df_v2["DEPT"].min())
print("Maximum:", df_v2["DEPT"].max())

# ------------------------------------------------------------
# Check depth ordering within each well
# ------------------------------------------------------------

depth_diff = (
    df_v2
    .groupby("WELL")["DEPT"]
    .diff()
)

print("\nDepth difference statistics:")
display(depth_diff.describe())

# ------------------------------------------------------------
# Check for problematic depth differences
# ------------------------------------------------------------

print("\nDepth differences <= 0:")
print((depth_diff <= 0).sum())

print("\nDepth differences > 0:")
print((depth_diff > 0).sum())

# ------------------------------------------------------------
# Show a few wells
# ------------------------------------------------------------

print("\nExample depth values:")
display(
    df_v2[["WELL", "DEPT"]]
    .head(20)
)
```

```python
# ============================================================
# CELL 8 — INSPECT DEPTH GAPS
# ============================================================

# Calculate depth differences within each well.
depth_gap = (
    df_v2
    .groupby("WELL")["DEPT"]
    .diff()
)

# Remove the first row of every well (NaN difference).
depth_gap = depth_gap.dropna()

print("==============================================")
print("DEPTH GAP ANALYSIS")
print("==============================================")

print("\nAll depth-gap statistics:")
display(depth_gap.describe())

# ------------------------------------------------------------
# Count different gap sizes
# ------------------------------------------------------------

print("\nGap <= 0.152 m:")
print((depth_gap <= 0.152001).sum())

print("\nGap > 0.152 m:")
print((depth_gap > 0.152001).sum())

print("\nGap > 0.304 m:")
print((depth_gap > 0.304001).sum())

print("\nGap > 1 m:")
print((depth_gap > 1.0).sum())

print("\nGap > 10 m:")
print((depth_gap > 10.0).sum())

print("\nGap > 100 m:")
print((depth_gap > 100.0).sum())

# ------------------------------------------------------------
# Show the largest gaps
# ------------------------------------------------------------

print("\nLargest 20 depth gaps:")

largest_gaps = (
    df_v2[["WELL", "DEPT"]]
    .copy()
)

largest_gaps["DEPTH_GAP"] = (
    largest_gaps
    .groupby("WELL")["DEPT"]
    .diff()
)

display(
    largest_gaps
    .sort_values("DEPTH_GAP", ascending=False)
    .head(20)
)
```

```python
# ============================================================
# CELL 9 — CALCULATE FULL-DEPTH GRADIENTS
# ============================================================

# These are the log curves for which we want depth gradients.
GRADIENT_CURVES = [
    "GR",
    "RDEP_LOG10",
    "RMED_LOG10",
    "DTC",
    "RHOB",
]

# Get the LAS files once.
las_files = sorted(RAW_DIR.glob("*.las"))

# This list will hold the gradient dataframe for each well.
gradient_parts = []

print("Calculating gradients from the full depth sequences...\n")


# ============================================================
# PROCESS EACH WELL
# ============================================================

for i, las_path in enumerate(las_files, start=1):

    # --------------------------------------------------------
    # 1. Read original LAS file
    # --------------------------------------------------------
    las = lasio.read(las_path)

    raw = las.df().reset_index()

    # Normalize all curve names to uppercase.
    raw.columns = [str(c).upper() for c in raw.columns]

    # --------------------------------------------------------
    # 2. Handle missing RMED
    # --------------------------------------------------------
    if "RMED" not in raw.columns:
        raw["RMED"] = np.nan

    # --------------------------------------------------------
    # 3. Replace LAS NULL values with NaN
    # --------------------------------------------------------
    raw = raw.replace([-999.25, -999.250000], np.nan)

    # --------------------------------------------------------
    # 4. Apply our established cleaning rules
    # --------------------------------------------------------

    # GR <= 0 is not valid for our preprocessing.
    raw.loc[raw["GR"] < 0, "GR"] = np.nan

    # Resistivity must be positive.
    raw.loc[raw["RDEP"] <= 0, "RDEP"] = np.nan
    raw.loc[raw["RMED"] <= 0, "RMED"] = np.nan

    # Sonic and density must be positive.
    raw.loc[raw["DTC"] <= 0, "DTC"] = np.nan
    raw.loc[raw["RHOB"] <= 0, "RHOB"] = np.nan

    # --------------------------------------------------------
    # 5. Create log10 resistivity
    # --------------------------------------------------------

    raw["RDEP_LOG10"] = np.log10(raw["RDEP"])
    raw["RMED_LOG10"] = np.log10(raw["RMED"])

    # --------------------------------------------------------
    # 6. Calculate gradients using the FULL depth sequence
    # --------------------------------------------------------

    depth = raw["DEPT"].to_numpy(dtype=float)

    gradient_data = {
        "DEPT": depth
    }

    for curve in GRADIENT_CURVES:

        values = raw[curve].to_numpy(dtype=float)

        # Start with all NaN.
        gradient = np.full(len(raw), np.nan, dtype=float)

        # Difference between consecutive depth samples.
        dz = depth[1:] - depth[:-1]

        # Difference between consecutive log values.
        dx = values[1:] - values[:-1]

        # Only calculate where both values and depth are valid.
        valid = (
            np.isfinite(values[:-1])
            & np.isfinite(values[1:])
            & np.isfinite(depth[:-1])
            & np.isfinite(depth[1:])
            & (dz > 0)
        )

        # Forward gradient:
        #
        # dX/dz = (X[i+1] - X[i]) / (z[i+1] - z[i])
        #
        gradient[:-1][valid] = dx[valid] / dz[valid]

        gradient_data[f"{curve}_GRAD"] = gradient

    # --------------------------------------------------------
    # 7. Recover the same WELL name used by our processed data
    # --------------------------------------------------------

    well_name = las_path.stem

    gradient_well = pd.DataFrame(gradient_data)

    gradient_well["WELL"] = well_name

    # Put columns in a predictable order.
    gradient_well = gradient_well[
        ["WELL", "DEPT"]
        + [f"{curve}_GRAD" for curve in GRADIENT_CURVES]
    ]

    gradient_parts.append(gradient_well)

    if i % 20 == 0 or i == len(las_files):
        print(f"Processed {i}/{len(las_files)} wells")


# ============================================================
# COMBINE ALL WELLS
# ============================================================

full_gradients = pd.concat(
    gradient_parts,
    ignore_index=True
)


# ============================================================
# SUMMARY
# ============================================================

print("\n==============================================")
print("FULL GRADIENT DATASET")
print("==============================================")

print("Rows :", len(full_gradients))
print("Wells:", full_gradients["WELL"].nunique())

print("\nColumns:")
print(full_gradients.columns.tolist())


# ============================================================
# GRADIENT MISSINGNESS
# ============================================================

gradient_columns = [
    f"{curve}_GRAD"
    for curve in GRADIENT_CURVES
]

print("\nGradient missingness (%):")

gradient_missingness = (
    full_gradients[gradient_columns]
    .isna()
    .mean()
    .mul(100)
    .round(3)
)

display(gradient_missingness)


# ============================================================
# GRADIENT STATISTICS
# ============================================================

print("\nGradient statistics:")

gradient_statistics = (
    full_gradients[gradient_columns]
    .describe()
    .T
)

display(gradient_statistics)
```

```python
# ============================================================
# CELL 10 — VALIDATE GRADIENT DEPTH SPACING
# ============================================================

# ------------------------------------------------------------
# 1. Calculate depth difference inside each well
# ------------------------------------------------------------

full_gradients["DZ"] = (
    full_gradients
    .groupby("WELL")["DEPT"]
    .diff()
)

print("==============================================")
print("FULL-DEPTH GRADIENT SPACING")
print("==============================================")

print("\nDepth-gap statistics:")
display(
    full_gradients["DZ"]
    .dropna()
    .describe()
)


# ------------------------------------------------------------
# 2. Count normal and abnormal depth steps
# ------------------------------------------------------------

normal_step = np.isclose(
    full_gradients["DZ"],
    0.152,
    atol=1e-6
)

abnormal_step = (
    full_gradients["DZ"].notna()
    & ~normal_step
)

print("\nNormal 0.152 m transitions:")
print(normal_step.sum())

print("\nAbnormal transitions:")
print(abnormal_step.sum())

print(
    "\nPercentage abnormal:",
    round(abnormal_step.mean() * 100, 5),
    "%"
)


# ------------------------------------------------------------
# 3. Show the largest abnormal gaps
# ------------------------------------------------------------

print("\nLargest depth gaps in full LAS sequences:")

display(
    full_gradients[
        ["WELL", "DEPT", "DZ"]
    ]
    .sort_values("DZ", ascending=False)
    .head(20)
)


# ------------------------------------------------------------
# 4. Invalidate gradients across abnormal depth gaps
# ------------------------------------------------------------

gradient_columns = [
    "GR_GRAD",
    "RDEP_LOG10_GRAD",
    "RMED_LOG10_GRAD",
    "DTC_GRAD",
    "RHOB_GRAD",
]

for column in gradient_columns:
    full_gradients.loc[abnormal_step, column] = np.nan


# ------------------------------------------------------------
# 5. Recalculate statistics after removing
#    gradients across abnormal depth gaps
# ------------------------------------------------------------

print("\n==============================================")
print("CLEANED GRADIENT STATISTICS")
print("==============================================")

gradient_statistics_clean = (
    full_gradients[gradient_columns]
    .describe()
    .T
)

display(gradient_statistics_clean)


# ------------------------------------------------------------
# 6. Missingness after depth-gap filtering
# ------------------------------------------------------------

print("\nGradient missingness after depth-gap filtering (%):")

gradient_missingness_clean = (
    full_gradients[gradient_columns]
    .isna()
    .mean()
    .mul(100)
    .round(3)
)

display(gradient_missingness_clean)
```

```python
# ============================================================
# CELL 11 — MERGE VERIFIED GRADIENTS INTO ML DATASET
# ============================================================

gradient_columns = [
    "GR_GRAD",
    "RDEP_LOG10_GRAD",
    "RMED_LOG10_GRAD",
    "DTC_GRAD",
    "RHOB_GRAD",
]

# ------------------------------------------------------------
# 1. Check that WELL + DEPT uniquely identifies a sample
# ------------------------------------------------------------

duplicate_gradient_keys = full_gradients.duplicated(
    subset=["WELL", "DEPT"]
).sum()

duplicate_ml_keys = df_v2.duplicated(
    subset=["WELL", "DEPT"]
).sum()

print("Duplicate (WELL, DEPT) keys in full gradients:",
      duplicate_gradient_keys)

print("Duplicate (WELL, DEPT) keys in ML dataset:",
      duplicate_ml_keys)

if duplicate_gradient_keys > 0:
    raise ValueError(
        "Duplicate WELL + DEPT found in full gradient dataset."
    )

if duplicate_ml_keys > 0:
    raise ValueError(
        "Duplicate WELL + DEPT found in ML dataset."
    )


# ------------------------------------------------------------
# 2. Keep only the gradient information we need
# ------------------------------------------------------------

gradient_lookup = full_gradients[
    ["WELL", "DEPT"] + gradient_columns
].copy()


# ------------------------------------------------------------
# 3. Merge gradients into our existing feature dataset
# ------------------------------------------------------------

df_v2_gradient = df_v2.merge(
    gradient_lookup,
    on=["WELL", "DEPT"],
    how="left",
    validate="one_to_one"
)


# ------------------------------------------------------------
# 4. Verify row count
# ------------------------------------------------------------

print("\n==============================================")
print("GRADIENT MERGE CHECK")
print("==============================================")

print("Original ML rows :", len(df_v2))
print("New V2 rows      :", len(df_v2_gradient))

if len(df_v2_gradient) != len(df_v2):
    raise ValueError(
        "Row count changed after gradient merge."
    )

print("✓ Row count preserved.")


# ------------------------------------------------------------
# 5. Check that every ML row received a gradient record
# ------------------------------------------------------------

print("\nGradient missingness after merge (%):")

gradient_missingness_v2 = (
    df_v2_gradient[gradient_columns]
    .isna()
    .mean()
    .mul(100)
    .round(3)
)

display(gradient_missingness_v2)


# ------------------------------------------------------------
# 6. Show final shape and columns
# ------------------------------------------------------------

print("\nOriginal feature count:", df.shape[1])

print(
    "Feature count after adding DEPT:",
    df_v2.shape[1]
)

print(
    "Feature count after adding gradients:",
    df_v2_gradient.shape[1]
)

print("\nNew gradient columns:")
print(gradient_columns)


# ------------------------------------------------------------
# 7. Inspect a few rows
# ------------------------------------------------------------

print("\nExample rows with gradients:")

display(
    df_v2_gradient[
        [
            "WELL",
            "DEPT",
            "GR",
            "DTC",
            "RHOB",
            "GR_GRAD",
            "DTC_GRAD",
            "RHOB_GRAD",
        ]
    ].head(15)
)
```

```python
# ============================================================
# CELL 12 — SAVE FEATURE SET V2
# ============================================================

V2_FILE = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "engineered_features_v2.parquet"
)

# ------------------------------------------------------------
# 1. Save the complete V2 dataset
# ------------------------------------------------------------

df_v2_gradient.to_parquet(
    V2_FILE,
    index=False
)

print("Saved Feature Set V2:")
print(V2_FILE)


# ------------------------------------------------------------
# 2. Reload the saved file
# ------------------------------------------------------------

v2_check = pd.read_parquet(V2_FILE)


# ------------------------------------------------------------
# 3. Verify the saved dataset
# ------------------------------------------------------------

print("\n==============================================")
print("FEATURE SET V2 VERIFICATION")
print("==============================================")

print("Original V2 shape :", df_v2_gradient.shape)
print("Reloaded V2 shape :", v2_check.shape)

print(
    "\nShape matches:",
    df_v2_gradient.shape == v2_check.shape
)

print(
    "Gradient columns present:",
    all(
        column in v2_check.columns
        for column in gradient_columns
    )
)

print(
    "DEPT present:",
    "DEPT" in v2_check.columns
)

print(
    "Wells:",
    v2_check["WELL"].nunique()
)

print(
    "Rows:",
    len(v2_check)
)


# ------------------------------------------------------------
# 4. Final V2 column summary
# ------------------------------------------------------------

print("\nNew gradient columns:")
print(gradient_columns)

print("\nFeature Set V2 columns:")
print(v2_check.columns.tolist())
```

Experiment Report --- Gradient Features and Tree-Based Models
1. Experiment Objective
This experiment investigated whether depth-based geological gradient
features improve lithology prediction when added to the existing W5
feature representation.
It also compared three tree-based learners under the same experimental
protocol:
- Extra Trees
- XGBoost
- LightGBM
The purpose was to determine whether gradients are useful, which learner
can exploit them most effectively, and whether difficult
lithologies---especially 65030 (Sandstone/Shale)---improve.
The official PUBLIC and BLIND_TEST wells were not used for model
development.
2. Dataset and Experimental Setup
The engineered dataset contained 1,430,974 labelled samples from 118
wells.
The official FORCE 2020 well split was preserved:
- 98 official TRAIN wells
- 10 PUBLIC wells
- 10 BLIND_TEST wells
Only the 98 official TRAIN wells were used for development.
Those 98 wells were divided using random_state=42 into:
- 78 DEV_TRAIN wells
- 20 DEV_VALIDATION wells
The development split was saved as:
data/splits/development_split.csv
The PUBLIC and BLIND_TEST wells remained untouched.
The split was performed at the well level, not at the
individual-sample level. This is important because nearby depth samples
from the same well are highly correlated. A random row split could
therefore make validation unrealistically easy.
3. Gradient Feature Construction
Five new gradient features were calculated:
- GR_GRAD
- RDEP_LOG10_GRAD
- RMED_LOG10_GRAD
- DTC_GRAD
- RHOB_GRAD
A gradient represents the rate of change of a log with depth:
\(`\frac{dX}{dz}`{=tex}\)
For example:
\(GR_GRAD `\approx `{=tex}`\frac{GR_{i+1}-GR_i}{z_{i+1}-z_i}`{=tex}\)
A positive gradient means the log is increasing with depth, while a
negative gradient means it is decreasing.
The gradients were calculated from the complete LAS depth sequences,
rather than from the already filtered labelled dataset.
The original LAS sequences were checked and showed regular approximately
0.152 m depth spacing:
- 2,337,521 normal transitions
- 0 abnormal transitions
Earlier large gaps in the labelled dataset were therefore mainly
consequences of filtering/removing rows during preprocessing, not gaps
in the original LAS measurements.
4. Why Gradients Can Help
A raw well-log value describes the property at a particular depth.
A gradient adds another piece of information:
How rapidly is that property changing with depth?

Lithological boundaries are often associated with changes in physical
properties. For example:
- GR gradients can indicate changes between cleaner and shalier
  intervals.
- DTC gradients can indicate changes in acoustic properties.
- RHOB gradients can indicate density contrasts.
- Resistivity gradients can indicate changes in formation/fluid
  properties.
Two samples with similar absolute log values can therefore have
different geological contexts if one is inside a stable interval and the
other is near a sharp transition.
Gradient features attempt to provide this local transition information
to the model.

WELL + DEPT

# 12. Meaning of Accuracy

Accuracy is:

\[ Accuracy = `\frac{\text{correct predictions}}`{=tex}
{`\text{total predictions}`{=tex}} \]

It is useful, but the dataset is strongly imbalanced. Shale 65000
represents a large fraction of the samples.

Therefore a model can have reasonable accuracy while performing poorly
on rare lithologies.

Accuracy should therefore not be the sole selection metric for this
project.

------------------------------------------------------------------------

# 13. Meaning of Balanced Accuracy

Balanced accuracy averages the recall of every class:

\[ Balanced Accuracy = `\frac{1}{K}`{=tex} `\sum`{=tex}\_{c=1}\^{K}
Recall_c \]

Every class therefore receives equal importance.

This is more informative than raw accuracy when rare lithologies matter.

The three gradient experiments produced balanced accuracies around 0.40,
showing some improvement in class-balanced performance but also
substantial remaining confusion.

------------------------------------------------------------------------

# 14. Meaning of F1

For an individual class:

\[ F1 = `\frac{2PR}{P+R}`{=tex} \]

where:

-   \(P\) = precision
-   \(R\) = recall

Precision asks:

> When the model predicts this lithology, how often is it correct?

Recall asks:

> Of all true samples of this lithology, how many did the model find?

F1 balances both.

This is particularly useful for rare lithologies because predicting a
rare class excessively can increase recall while destroying precision.

------------------------------------------------------------------------

# 15. Macro F1

Macro F1 calculates the F1 score separately for every class and gives
every class equal importance:

\[ MacroF1 = `\frac{1}{K}`{=tex} `\sum`{=tex}\_{c=1}\^{K}F1_c \]

This is one of the most important development metrics for this project
because it prevents the dominant Shale class from hiding poor rare-class
performance.

The best Macro F1 in this experiment was produced by XGBoost:

**0.369926**

LightGBM produced:

**0.347732**

Extra Trees produced:

**0.302777**

------------------------------------------------------------------------

# 16. Weighted F1

Weighted F1 weights each class according to its sample count:

\[ WeightedF1 = `\sum`{=tex}\_c `\frac{N_c}{N}`{=tex}F1_c \]

It is therefore strongly influenced by common lithologies.

A high Weighted F1 can coexist with poor rare-class performance.

It should therefore be treated as a secondary metric rather than the
primary rare-class selection metric.

------------------------------------------------------------------------

# 17. Why XGBoost Performed Best Overall

Extra Trees constructs many randomized trees independently and averages
them.

Boosting methods construct trees sequentially, with later trees
attempting to correct errors made by previous trees.

This sequential error-correction mechanism can be useful when lithology
classes overlap strongly in feature space.

The experiment supports the hypothesis that boosted trees can exploit
the combination of:

-   raw log values,
-   rolling statistics,
-   missingness indicators,
-   resistivity separation,
-   depth gradients.

This does not prove that XGBoost will remain the best model after
additional feature engineering, but it makes XGBoost a strong candidate
for later experiments.

------------------------------------------------------------------------

# 18. Why LightGBM Remains Important

Although XGBoost had the highest overall Macro F1, LightGBM performed
better specifically on the main difficult target, 65030:

\[ F1\_{65030}=0.341241 \]

compared with approximately 0.30 for XGBoost.

Therefore LightGBM should not be discarded.

The eventual model selection should consider:

-   Macro F1,
-   Balanced Accuracy,
-   65030 performance,
-   rare-class performance,
-   FORCE penalty score,
-   generalization to unseen wells.

# 19. Current Scientific Conclusions

### Gradient features are technically valid

### Gradients alone do not solve the difficult classes

Adding five gradients to Extra Trees did not produce a decisive

65030, Chalk, Dolomite, and Halite remain difficult.

### Boosted trees are promising

Both XGBoost and LightGBM performed better than Extra Trees in Macro F1.

XGBoost achieved the highest overall Macro F1:

**0.369926**

LightGBM achieved the strongest 65030 F1:

**0.341241**

This supports continuing development with boosted-tree models.

# 20. Important Limitations

These are **development-validation results**, not final blind results.

The 10 official BLIND_TEST wells have not been used for tuning based on
these experiments.

The Basement class cannot be evaluated on this particular development
validation split because no Basement sample is present in
DEV_VALIDATION.

The class-weighting strategy can also produce undesirable
precision/recall trade-offs for extremely rare classes. For example,
high recall can occur alongside very low precision.

Future experiments should therefore examine both class weighting and
probability/decision behavior.
